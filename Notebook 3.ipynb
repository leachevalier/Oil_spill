```

"""
VERSION: Starting day: 20th November 2015

IMPORTS
"""
from netCDF4 import Dataset
import numpy as np
import matplotlib.pyplot as plt
import os
import glob  # Automatically find all the NetCDF files
import pandas as pd

#utilisation de CARTOPY

import cartopy.crs as ccrs
import cartopy.feature as cfeature

os.makedirs("/work/figures2", exist_ok=True) #création de files pour enregistrer toutes les images


"""
LOAD FILES
"""

# Path to folder containing uploaded files in deepnote
path_data = "/work/"

# Load ensemble files sorted from 1 to 10
sic_files = sorted(glob.glob(path_data + "siconc*.nc"))  # files starting with siconc
u_files = sorted(glob.glob(path_data + "sivelu*.nc"))
v_files = sorted(glob.glob(path_data + "sivelv*.nc"))

print(os.listdir("/work/"))  # Only to see what is in the folder


"""
LOAD EASE GRID
"""

# Load EASE grid
grid = Dataset(path_data + "EASEGRID_x_y_lat_lon.nc")

XX = grid.variables["XX"][:]
YY = grid.variables["YY"][:]
LAT = grid.variables["LAT"][:]
LON = grid.variables["LON"][:]

grid_size_m = 25000  # Size of the grid (in meters)


"""
PARAMETERS
"""

# Starting points of oil spill
x0, y0 = 180, 200

# Number of oil beads released: 100
N = 100

# One day = 24 hours = 86,400 seconds
# Seconds because the velocity files are in meters per second.
dt = 86400

# Limit of quantity of ice for oil beads to be considered still in ice
sic_threshold = 0.15

# Starting day of the oil spill: 
# For NEMO Day 0 = June 1st but 1st day is November 20th = day 172
START_DAY = 172


"""
IDW INTERPOLATION FUNCTION
"""

def idw(position_x, position_y, field):

    # Take lower integer coordinates around the particle
    x_int = int(np.floor(position_x))
    y_int = int(np.floor(position_y))

    # 4 surrounding grid points
    neighbors = [
        (x_int, y_int),
        (x_int + 1, y_int),
        (x_int, y_int + 1),
        (x_int + 1, y_int + 1)
    ]

    weights = []
    values = []

    # Loop through neighboring points
    for nx, ny in neighbors:

        # Check if point is inside the grid
        if 0 <= nx < field.shape[1] and 0 <= ny < field.shape[0]:

            # Distance² between particle and grid point
            # +1e-12 avoids division by zero
            dist2 = (
                (position_x - nx)**2
                + (position_y - ny)**2
                + 1e-12
            )

            # IDW weight = 1 / distance²
            weights.append(1 / dist2)

            # Value of field at neighboring point
            values.append(field[ny, nx])

    # Weighted average
    return (
        np.sum(np.array(weights) * np.array(values))
        / np.sum(weights)
    )



# =========================
# GLOBAL PROBABILITY MAP
# =========================

# Same size as sea ice grid
probability_map = np.zeros_like(LON, dtype=float)

# Store statistics for each ensemble
mean_melt_time_all = []
mean_distance_all = []
membre_labels = []
southest_latitude_all = []

"""
LOOP OVER THE 10 MEMBERS
"""

for k in range(10):

    print("Running membre", k + 1)

    """
    OPEN NETCDF FILES
    """

    # Open NetCDF files
    sic_ds = Dataset(sic_files[k])
    u_ds = Dataset(u_files[k])
    v_ds = Dataset(v_files[k])

    # Extract variables
    sic = sic_ds.variables["siconc"]
    u = u_ds.variables["sivelu"]
    v = v_ds.variables["sivelv"]

    # number of days -> for 2024: nt = 366
    number_of_days = sic.shape[0]
    """
    OIL SPILL OF 100 OIL BEADS
    """

    # Make random result reproducible
    np.random.seed(42)

    # Oil beads start from the same starting point
    # but with a small position deviance
    x = x0 + np.random.uniform(0, 1, N)
    y = y0 + np.random.uniform(0, 1, N)
    

    # Save trajectories
    traj_x = [x.copy()]
    traj_y = [y.copy()]

    # =========================
    # STATISTICS VARIABLES
    # =========================

    # Time before melting
    melt_time = np.full(N, np.nan)

    # Total distance traveled
    distance_traveled = np.zeros(N)

    """
    TRAJECTORY OF OIL BEADS
    """
    
    for t in range(START_DAY,number_of_days - 1):

        # Variables for current day
        u_t = u[t]

        # IMPORTANT:
        # keep original direction of v
        v_t = v[t]

        sic_t = sic[t]


        """
        LOOP OVER OIL BEADS
        """

        for i in range(N):

            # Ignore removed particles
            if np.isnan(x[i]) or np.isnan(y[i]):
                continue


            """
            INTERPOLATE VELOCITY
            """

            # Interpolate velocity at particle position
            ui = idw(x[i], y[i], u_t)
            vi = idw(x[i], y[i], v_t)

            # Remove invalid interpolation results
            if np.isnan(ui) or np.isnan(vi):
                 x[i] = np.nan
                 y[i] = np.nan
                 continue


            """
            MOVE PARTICLE
            """
            # Save old position
            old_x = x[i]
            old_y = y[i]

            # Move particle during one day
            x[i] += ui * dt / grid_size_m
            y[i] += vi * dt / grid_size_m

            # Compute traveled distance in grid cells
            dx = x[i] - old_x
            dy = y[i] - old_y

            # Convert to meters
            distance_step = np.sqrt(dx**2 + dy**2) * grid_size_m

            # Add to total distance
            distance_traveled[i] += distance_step

            # Safety check after movement
            if np.isnan(x[i]) or np.isnan(y[i]):
                continue

            """
            CHECK IF PARTICLE IS STILL INSIDE GRID
            """

            ny, nx = sic_t.shape

            xi = int(round(x[i]))
            yi = int(round(y[i]))

            # safety check
            if (
                xi < 0
                or yi < 0
                or xi >= nx
                or yi >= ny
            ):

                x[i] = np.nan
                y[i] = np.nan

                continue


            """
            CHECK SEA ICE CONCENTRATION
            """

            sic_i = idw(x[i], y[i], sic_t)

            # Remove particles outside sea ice
            """if sic_i < sic_threshold:
                # Save melt time ONLY the first time
                if np.isnan(x[i]) == False:
                   melt_time[i] = t

                x[i] = np.nan
                y[i] = np.nan"""

            if sic_i < sic_threshold:

                # Save melt time only once
                if np.isnan(melt_time[i]):

                    melt_time[i] = t - START_DAY

                # Remove particle
                x[i] = np.nan
                y[i] = np.nan

                continue


        """
        SAVE DAILY POSITIONS
        """

        traj_x.append(x.copy())
        traj_y.append(y.copy())

        # =========================
        # UPDATE PROBABILITY MAP
        # =========================

        for i in range(N):

            if np.isnan(x[i]) or np.isnan(y[i]):
                continue

            xi = int(round(x[i]))
            yi = int(round(y[i]))

            if (
                0 <= xi < probability_map.shape[1]
                and 0 <= yi < probability_map.shape[0]
            ):

                # Add one "hit"
                probability_map[yi, xi] += 1

        """
        STOP IF ALL PARTICLES REMOVED
        """

        if np.all(np.isnan(x)):
            break

    # =========================
    # FINALIZE MELT TIMES
    # =========================

    # Particles that never melted
    # survived until end of simulation

    melt_time[np.isnan(melt_time)] = number_of_days - START_DAY

    """
    CONVERT TO NUMPY ARRAYS
    """

    traj_x = np.array(traj_x)
    traj_y = np.array(traj_y)

    
    """
    PLOT TRAJECTORIES
    """

    plt.figure(figsize=(8, 8))

    # boucle sur les trajectoires des particules
    for i in range(traj_x.shape[1]):

        plt.plot(
            traj_x[:, i],
            traj_y[:, i],
            alpha=0.4
        )

    plt.gca().invert_yaxis()  # important

    plt.xlabel("X")
    plt.ylabel("Y")

    plt.title(f"Oil bead trajectories - Ensemble Membre {k+1}")

    plt.grid(True)

    plt.savefig(
        f"/work/figures2/trajectory_grid_membre_{k+1}.png",
        dpi=300
    )

    plt.show()
    plt.close()


    """
    TRAJECTOIRE SUR CARTE
    """

    # tableaux vides pour stocker les coordonnées géographiques
    traj_lon = np.full(traj_x.shape, np.nan)
    traj_lat = np.full(traj_y.shape, np.nan)

    # conversion grille -> latitude / longitude
    for t in range(traj_x.shape[0]):

        for i in range(traj_x.shape[1]):

            if np.isnan(traj_x[t, i]) or np.isnan(traj_y[t, i]):
                continue

            xi = int(round(traj_x[t, i]))
            yi = int(round(traj_y[t, i]))

            if (
                0 <= xi < LON.shape[1]
                and 0 <= yi < LON.shape[0]
            ):

                traj_lon[t, i] = LON[yi, xi]
                traj_lat[t, i] = LAT[yi, xi]


    """
    CARTE POLAIRE
    """

    fig = plt.figure(figsize=(10, 10))

    ax = plt.axes(projection=ccrs.NorthPolarStereo())

    mean_sic = np.nanmean(sic, axis=0)

    """
    FOND : SEA ICE CONCENTRATION
    """
    
    pcm = ax.pcolormesh(
        LON,
        LAT,
        mean_sic,
        transform=ccrs.PlateCarree(),
        cmap="Blues_r",
        shading="auto",
        vmin=0,
        vmax=1
    )

    plt.colorbar(
        pcm,
        ax=ax,
        shrink=0.7,
        label="Sea ice concentration (%)"
    )

    """
    AJOUT DES CONTINENTS
    """

    ax.coastlines()

    ax.add_feature(
        cfeature.LAND,
        facecolor="lightgray"
    )

    ax.add_feature(
        cfeature.OCEAN,
        facecolor="white"
    )

    # limite Arctique
    ax.set_extent(
        [-180, 180, 60, 90],
        crs=ccrs.PlateCarree()
    )


    """
    TRAJECTOIRES of beads in red on map with continents
    """

    for i in range(traj_lon.shape[1]):

        lon_i = traj_lon[:, i]
        lat_i = traj_lat[:, i]

        mask = (
            ~np.isnan(lon_i)
            & ~np.isnan(lat_i)
        )

        if np.any(mask):

            ax.plot(
                lon_i[mask],
                lat_i[mask],
                transform=ccrs.PlateCarree(),
                color="red",
                alpha=0.4,
                linewidth=1
            )


    """
    POINT DE DEPART
    """

    ax.scatter(
        LON[y0, x0],
        LAT[y0, x0],
        color="red",
        s=50,
        transform=ccrs.PlateCarree()
    )

    plt.title(
        f"Trajectoires des particules sur la carte - Membre {k+1}"
    )

    # =========================
    # MEMBRES STATISTICS
    # =========================

    # Mean melt time
    mean_melt_time = np.nanmean(melt_time)

    # Mean distance traveled (km)
    mean_distance = np.nanmean(distance_traveled) / 1000

    #Furthest point touched by oil beads
    southest_latitude = np.nanmin(traj_lat)

    # Save statistics
    mean_melt_time_all.append(mean_melt_time)
    mean_distance_all.append(mean_distance)
    southest_latitude_all.append(southest_latitude)
    membre_labels.append(f"E{k+1}")

    print("Mean melt time =", mean_melt_time, "days")
    print("Mean distance traveled =", mean_distance, "km")
    print("Southest latitude =", southest_latitude, "°N")

    plt.savefig(
    f"/work/figures2/map_membre_{k+1}.png",
    dpi=300
    )

    plt.show()
    plt.close()
    # Close NetCDF files for this ensemble member
    sic_ds.close()
    u_ds.close()
    v_ds.close()


"""
FINAL PROBABILITY MAP
"""

# Divide by number of ensembles
probability_map = probability_map / 10

fig = plt.figure(figsize=(10,10))

ax = plt.axes(projection=ccrs.NorthPolarStereo())

pcm = ax.pcolormesh(
    LON,
    LAT,
    probability_map,
    transform=ccrs.PlateCarree(),
    cmap="hot",
    shading="auto"
)

plt.colorbar(
    pcm,
    ax=ax,
    label="Probability of oil spill impact"
)

ax.coastlines()

ax.add_feature(
    cfeature.LAND,
    facecolor="lightgray"
)

ax.set_extent(
    [-180, 180, 60, 90],
    crs=ccrs.PlateCarree()
)

plt.title("Probability map of oil spill contamination")

plt.show()
plt.close()

"""
HISTOGRAM : MEAN MELT TIME
"""

plt.figure(figsize=(10,6))

plt.bar(
    membre_labels,
    mean_melt_time_all,
    color="royalblue", 
    edgecolor="black"
)

plt.xlabel("Membre year")
plt.ylabel("Mean melt time (days)")

plt.title("Mean time before oil beads melt")

plt.grid(axis="y", alpha=0.3)

plt.savefig(
    "/work/figures2/melt_time_histogram.png",
    dpi=300
)

plt.show()
plt.close()


"""
HISTOGRAM : MEAN DISTANCE
"""

plt.figure(figsize=(10,6))

plt.bar(
    membre_labels,
    mean_distance_all,
    color="deepskyblue", 
    edgecolor="black"
)

plt.xlabel("Membre year")
plt.ylabel("Mean traveled distance (km)")

plt.title("Mean traveled distance of oil beads")

plt.grid(axis="y", alpha=0.3)

plt.savefig(
    "/work/figures2/distance_histogram.png",
    dpi=300
)

plt.show()
plt.close()
```

```
"""
VERSION: Starting day: 20th November 2015 with connection between ensemble members 

IMPORTS
"""
from netCDF4 import Dataset
import numpy as np
import matplotlib.pyplot as plt
import os
import glob  # Automatically find all the NetCDF files
import pandas as pd

#utilisation de CARTOPY

import cartopy.crs as ccrs
import cartopy.feature as cfeature

os.makedirs("/work/figures3", exist_ok=True) #création de files pour enregistrer toutes les images


"""
LOAD FILES
"""

# Path to folder containing uploaded files in deepnote
path_data = "/work/"

# Load ensemble files sorted from 1 to 10
sic_files = sorted(glob.glob(path_data + "siconc*.nc"))  # files starting with siconc
u_files = sorted(glob.glob(path_data + "sivelu*.nc"))
v_files = sorted(glob.glob(path_data + "sivelv*.nc"))

print(os.listdir("/work/"))  # Only to see what is in the folder


"""
LOAD EASE GRID
"""

# Load EASE grid
grid = Dataset(path_data + "EASEGRID_x_y_lat_lon.nc")

XX = grid.variables["XX"][:]
YY = grid.variables["YY"][:]
LAT = grid.variables["LAT"][:]
LON = grid.variables["LON"][:]

grid_size_m = 25000  # Size of the grid (in meters)


"""
PARAMETERS
"""

# Starting points of oil spill
x0, y0 = 180, 200

# Number of oil beads released: 100
N = 100

# One day = 24 hours = 86,400 seconds
# Seconds because the velocity files are in meters per second.
dt = 86400

# Limit of quantity of ice for oil beads to be considered still in ice
sic_threshold = 0.15

# Starting day of the oil spill: 
# For NEMO Day 0 = June 1st but 1st day is November 20th = day 172
START_DAY = 172

# Number of ensemble members that can be used to continue the simulation
# 2 = original member + next member
# 10 = use all members if needed
MAX_MEMBERS_TO_CHAIN = 10


"""
IDW INTERPOLATION FUNCTION
"""

def idw(position_x, position_y, field):

    # Take lower integer coordinates around the particle
    x_int = int(np.floor(position_x))
    y_int = int(np.floor(position_y))

    # 4 surrounding grid points
    neighbors = [
        (x_int, y_int),
        (x_int + 1, y_int),
        (x_int, y_int + 1),
        (x_int + 1, y_int + 1)
    ]

    weights = []
    values = []

    # Loop through neighboring points
    for nx, ny in neighbors:

        # Check if point is inside the grid
        if 0 <= nx < field.shape[1] and 0 <= ny < field.shape[0]:

            # Distance² between particle and grid point
            # +1e-12 avoids division by zero
            dist2 = (
                (position_x - nx)**2
                + (position_y - ny)**2
                + 1e-12
            )

            # IDW weight = 1 / distance²
            weights.append(1 / dist2)

            # Value of field at neighboring point
            values.append(field[ny, nx])

    # Weighted average
    return (
        np.sum(np.array(weights) * np.array(values))
        / np.sum(weights)
    )



# =========================
# GLOBAL PROBABILITY MAP
# =========================

# Same size as sea ice grid
probability_map = np.zeros_like(LON, dtype=float)

# Store statistics for each ensemble
mean_melt_time_all = []
mean_distance_all = []
membre_labels = []
southest_latitude_all = []

"""
LOOP OVER THE 10 MEMBERS
"""

for k in range(10):

    print("Running membre", k + 1)

    """
    OIL SPILL OF 100 OIL BEADS
    """

    # Make random result reproducible
    np.random.seed(42)

    # Oil beads start from the same starting point
    # but with a small position deviance
    x = x0 + np.random.uniform(0, 1, N)
    y = y0 + np.random.uniform(0, 1, N)
    

    # Save trajectories
    traj_x = [x.copy()]
    traj_y = [y.copy()]

    # Save which part of the trajectory is used
    # 0 = original member
    # 1, 2, ... = continuation using following members
    traj_chain = [0]

    # Total number of simulated days since the oil spill release
    elapsed_days = 0

    # This will store the sea ice concentration used as background for the map
    sic_background = None

    # Map to know if each cell has been touched by this ensemble member
    impact_this_member = np.zeros_like(LON, dtype=bool)

    # =========================
    # STATISTICS VARIABLES
    # =========================

    # Time before melting
    melt_time = np.full(N, np.nan)

    # Total distance traveled
    distance_traveled = np.zeros(N)

    """
    TRAJECTORY OF OIL BEADS WITH CONTINUATION INTO NEXT MEMBERS
    """

    # We can continue the trajectory using the next ensemble members
    for chain in range(MAX_MEMBERS_TO_CHAIN):

        # Select which member is used for this part of the trajectory
        # Example:
        # if k = 0 and chain = 0 -> member 1
        # if k = 0 and chain = 1 -> member 2
        # if k = 0 and chain = 2 -> member 3
        member_index = (k + chain) % 10

        print("Using data from membre", member_index + 1, "for trajectory of membre", k + 1)

        """
        OPEN NETCDF FILES
        """

        # Open NetCDF files
        sic_ds = Dataset(sic_files[member_index])
        u_ds = Dataset(u_files[member_index])
        v_ds = Dataset(v_files[member_index])

        # Extract variables
        sic = sic_ds.variables["siconc"]
        u = u_ds.variables["sivelu"]
        v = v_ds.variables["sivelv"]

        # number of days -> for 2024: nt = 366
        number_of_days = sic.shape[0]

        # For the original member, start on November 20th.
        # For the following members, start on June 1st.
        if chain == 0:
            first_day = START_DAY

            # Save background sea ice concentration for the map
            sic_background = np.array(sic[START_DAY])

        else:
            first_day = 0

        """
        TRAJECTORY OF OIL BEADS
        """
        
        for t in range(first_day, number_of_days - 1):

            # Variables for current day
            u_t = u[t]

            # IMPORTANT:
            # keep original direction of v
            v_t = v[t]

            sic_t = sic[t]


            """
            LOOP OVER OIL BEADS
            """

            for i in range(N):

                # Ignore removed particles
                if np.isnan(x[i]) or np.isnan(y[i]):
                    continue


                """
                INTERPOLATE VELOCITY
                """

                # Interpolate velocity at particle position
                ui = idw(x[i], y[i], u_t)
                vi = idw(x[i], y[i], v_t)

                # Remove invalid interpolation results
                if np.isnan(ui) or np.isnan(vi):
                    x[i] = np.nan
                    y[i] = np.nan
                    continue


                """
                MOVE PARTICLE
                """
                # Save old position
                old_x = x[i]
                old_y = y[i]

                # Move particle during one day
                x[i] += ui * dt / grid_size_m
                y[i] += vi * dt / grid_size_m

                # Compute traveled distance in grid cells
                dx = x[i] - old_x
                dy = y[i] - old_y

                # Convert to meters
                distance_step = np.sqrt(dx**2 + dy**2) * grid_size_m

                # Add to total distance
                distance_traveled[i] += distance_step

                # Safety check after movement
                if np.isnan(x[i]) or np.isnan(y[i]):
                    continue

                """
                CHECK IF PARTICLE IS STILL INSIDE GRID
                """

                ny, nx = sic_t.shape

                xi = int(round(x[i]))
                yi = int(round(y[i]))

                # safety check
                if (
                    xi < 0
                    or yi < 0
                    or xi >= nx
                    or yi >= ny
                ):

                    x[i] = np.nan
                    y[i] = np.nan

                    continue


                """
                CHECK SEA ICE CONCENTRATION
                """

                sic_i = idw(x[i], y[i], sic_t)

                # Remove particles outside sea ice
                """if sic_i < sic_threshold:
                    # Save melt time ONLY the first time
                    if np.isnan(x[i]) == False:
                       melt_time[i] = t

                    x[i] = np.nan
                    y[i] = np.nan"""

                if sic_i < sic_threshold:

                    # Save melt time only once
                    if np.isnan(melt_time[i]):

                        # Here we use elapsed_days because the simulation can continue
                        # beyond the first member.
                        melt_time[i] = elapsed_days + 1

                    # Remove particle
                    x[i] = np.nan
                    y[i] = np.nan

                    continue


            """
            SAVE DAILY POSITIONS
            """

            traj_x.append(x.copy())
            traj_y.append(y.copy())
            traj_chain.append(chain)

            # One more simulated day has passed
            elapsed_days += 1

            # =========================
            # UPDATE PROBABILITY MAP
            # =========================

            for i in range(N):

                if np.isnan(x[i]) or np.isnan(y[i]):
                    continue

                xi = int(round(x[i]))
                yi = int(round(y[i]))

                if (
                    0 <= xi < probability_map.shape[1]
                    and 0 <= yi < probability_map.shape[0]
                ):

                    # Add one "hit"
                    impact_this_member[yi, xi] = True

            """
            STOP IF ALL PARTICLES REMOVED
            """

            if np.all(np.isnan(x)):
                break

        # Close NetCDF files for this part of the trajectory
        sic_ds.close()
        u_ds.close()
        v_ds.close()

        # If all particles melted, stop using additional members
        if np.all(np.isnan(x)):
            break

    # Add this member impact to global probability map
    probability_map += impact_this_member.astype(float)

    # =========================
    # FINALIZE MELT TIMES
    # =========================

    # Particles that never melted
    # survived until end of simulation

    melt_time[np.isnan(melt_time)] = elapsed_days

    """
    CONVERT TO NUMPY ARRAYS
    """

    traj_x = np.array(traj_x)
    traj_y = np.array(traj_y)
    traj_chain = np.array(traj_chain)

    
    """
    PLOT TRAJECTORIES
    """

    plt.figure(figsize=(8, 8))

    # To avoid repeating labels 100 times in the legend
    label_red_done_grid = False
    label_yellow_done_grid = False

    # boucle sur les trajectoires des particules
    for i in range(traj_x.shape[1]):

        # We plot the trajectory segment by segment
        for tt in range(1, traj_x.shape[0]):

            # Skip if one of the two points is NaN
            if (
                np.isnan(traj_x[tt - 1, i])
                or np.isnan(traj_x[tt, i])
                or np.isnan(traj_y[tt - 1, i])
                or np.isnan(traj_y[tt, i])
            ):
                continue

            # If chain = 0, the trajectory is from the original member
            if traj_chain[tt] == 0:
                color = "red"
                label = "Original member" if not label_red_done_grid else None
                label_red_done_grid = True

            # If chain > 0, the trajectory continues using another member
            else:
                color = "yellow"
                label = "Extended trajectory" if not label_yellow_done_grid else None
                label_yellow_done_grid = True

            plt.plot(
                [traj_x[tt - 1, i], traj_x[tt, i]],
                [traj_y[tt - 1, i], traj_y[tt, i]],
                color=color,
                alpha=0.5,
                linewidth=1,
                label=label
            )

    plt.gca().invert_yaxis()  # important

    plt.xlabel("X")
    plt.ylabel("Y")

    plt.title(f"Oil bead trajectories - Ensemble Membre {k+1}")

    plt.grid(True)
    plt.legend()
    
    plt.savefig(
        f"/work/figures3/trajectory_grid_membre_{k+1}.png",
        dpi=300
    )

    plt.show()
    plt.close()
    

    """
    TRAJECTOIRE SUR CARTE
    """

    # tableaux vides pour stocker les coordonnées géographiques
    traj_lon = np.full(traj_x.shape, np.nan)
    traj_lat = np.full(traj_y.shape, np.nan)

    # conversion grille -> latitude / longitude
    for t in range(traj_x.shape[0]):

        for i in range(traj_x.shape[1]):

            if np.isnan(traj_x[t, i]) or np.isnan(traj_y[t, i]):
                continue

            xi = int(round(traj_x[t, i]))
            yi = int(round(traj_y[t, i]))

            if (
                0 <= xi < LON.shape[1]
                and 0 <= yi < LON.shape[0]
            ):

                traj_lon[t, i] = LON[yi, xi]
                traj_lat[t, i] = LAT[yi, xi]


    """
    CARTE POLAIRE
    """

    fig = plt.figure(figsize=(10, 10))

    ax = plt.axes(projection=ccrs.NorthPolarStereo())

    mean_sic = sic_background

    """
    FOND : SEA ICE CONCENTRATION
    """
    
    pcm = ax.pcolormesh(
        LON,
        LAT,
        mean_sic,
        transform=ccrs.PlateCarree(),
        cmap="Blues_r",
        shading="auto",
        vmin=0,
        vmax=1
    )

    plt.colorbar(
        pcm,
        ax=ax,
        shrink=0.7,
        label="Sea ice concentration (%)"
    )

    """
    AJOUT DES CONTINENTS
    """

    ax.coastlines()

    ax.add_feature(
        cfeature.LAND,
        facecolor="lightgray"
    )

    ax.add_feature(
        cfeature.OCEAN,
        facecolor="white"
    )

    # limite Arctique
    ax.set_extent(
        [-180, 180, 60, 90],
        crs=ccrs.PlateCarree()
    )


    """
    TRAJECTORIES of beads in red on map with continents
    """

    # To avoid repeating labels 100 times in the legend
    label_red_done_map = False
    label_yellow_done_map = False

    for i in range(traj_lon.shape[1]):

        # We plot the trajectory segment by segment
        for tt in range(1, traj_lon.shape[0]):

            lon_segment = [traj_lon[tt - 1, i], traj_lon[tt, i]]
            lat_segment = [traj_lat[tt - 1, i], traj_lat[tt, i]]

            # Skip if one of the two points is NaN
            if (
                np.isnan(lon_segment[0])
                or np.isnan(lon_segment[1])
                or np.isnan(lat_segment[0])
                or np.isnan(lat_segment[1])
            ):
                continue

            # If chain = 0, the trajectory is from the original member
            if traj_chain[tt] == 0:
                color = "red"
                label = "Original member" if not label_red_done_map else None
                label_red_done_map = True

            # If chain > 0, the trajectory continues using another member
            else:
                color = "yellow"
                label = "Extended trajectory" if not label_yellow_done_map else None
                label_yellow_done_map = True

            ax.plot(
                lon_segment,
                lat_segment,
                transform=ccrs.PlateCarree(),
                color=color,
                alpha=0.6,
                linewidth=1,
                label=label
            )


    """
    POINT DE DEPART
    """

    ax.scatter(
        LON[y0, x0],
        LAT[y0, x0],
        color="red",
        s=50,
        transform=ccrs.PlateCarree()
    )

    ax.legend()

    plt.title(
        f"Trajectories of oild beads on map - Ensemble membre {k+1}"
    )

    # =========================
    # MEMBRES STATISTICS
    # =========================

    # Mean melt time
    mean_melt_time = np.nanmean(melt_time)

    # Mean distance traveled (km)
    mean_distance = np.nanmean(distance_traveled) / 1000

    #Furthest point touched by oil beads
    southest_latitude = np.nanmin(traj_lat)

    # Save statistics
    mean_melt_time_all.append(mean_melt_time)
    mean_distance_all.append(mean_distance)
    southest_latitude_all.append(southest_latitude)
    membre_labels.append(f"E{k+1}")

    print("Mean melt time =", mean_melt_time, "days")
    print("Mean distance traveled =", mean_distance, "km")
    print("Southest latitude =", southest_latitude, "°N")

    plt.savefig(
    f"/work/figures3/map_membre_{k+1}.png",
    dpi=300
    )

    plt.show()
    plt.close()

"""
FINAL PROBABILITY MAP
"""

# Divide by number of ensembles
probability_map = probability_map / 10

fig = plt.figure(figsize=(10,10))

ax = plt.axes(projection=ccrs.NorthPolarStereo())

pcm = ax.pcolormesh(
    LON,
    LAT,
    probability_map,
    transform=ccrs.PlateCarree(),
    cmap="hot",
    shading="auto",
    vmin=0,
    vmax=1
)

plt.colorbar(
    pcm,
    ax=ax,
    label="Probability of oil spill impact"
)

ax.coastlines()

ax.add_feature(
    cfeature.LAND,
    facecolor="lightgray"
)

ax.set_extent(
    [-180, 180, 60, 90],
    crs=ccrs.PlateCarree()
)

plt.title("Probability map of oil spill contamination")

plt.savefig(
    "/work/figures3/probability_map.png",
    dpi=300
)

plt.show()
plt.close()


"""
HISTOGRAM : MEAN MELT TIME
"""

plt.figure(figsize=(10,6))

plt.bar(
    membre_labels,
    mean_melt_time_all,
    color="royalblue", 
    edgecolor="black"
)

plt.xlabel("Membre year")
plt.ylabel("Mean melt time (days)")

plt.title("Mean time before oil beads melt")

plt.grid(axis="y", alpha=0.3)

plt.savefig(
    "/work/figures3/melt_time_histogram.png",
    dpi=300
)

plt.show()
plt.close()


"""
HISTOGRAM : MEAN DISTANCE
"""

plt.figure(figsize=(10,6))

plt.bar(
    membre_labels,
    mean_distance_all,
    color="deepskyblue", 
    edgecolor="black"
)

plt.xlabel("Membre year")
plt.ylabel("Mean traveled distance (km)")

plt.title("Mean traveled distance of oil beads")

plt.grid(axis="y", alpha=0.3)

plt.savefig(
    "/work/figures3/distance_histogram.png",
    dpi=300
)

plt.show()
plt.close()
```

```
"""
PROBABILISTIC OIL SPILL IMPACT MAP
"""

"""
IMPORTS
"""

from netCDF4 import Dataset
import numpy as np
import matplotlib.pyplot as plt
import os
import glob

import cartopy.crs as ccrs
import cartopy.feature as cfeature


"""
CREATE FIGURE FOLDER
"""

os.makedirs("/work/figures_probability", exist_ok=True)


"""
LOAD FILES
"""

# Path to folder containing uploaded files in deepnote
path_data = "/work/"

# Load ensemble files sorted from 1 to 10
sic_files = sorted(glob.glob(path_data + "siconc*.nc"))
u_files = sorted(glob.glob(path_data + "sivelu*.nc"))
v_files = sorted(glob.glob(path_data + "sivelv*.nc"))

print("Sea ice concentration files:", len(sic_files))
print("Sea ice velocity u files:", len(u_files))
print("Sea ice velocity v files:", len(v_files))


"""
LOAD EASE GRID
"""

# Load EASE grid
grid = Dataset(path_data + "EASEGRID_x_y_lat_lon.nc")

XX = grid.variables["XX"][:]
YY = grid.variables["YY"][:]
LAT = grid.variables["LAT"][:]
LON = grid.variables["LON"][:]

grid_size_m = 25000  # Size of one grid cell in meters


"""
PARAMETERS
"""

# Starting point of oil spill
x0, y0 = 180, 200

# Number of oil beads released
N = 100

# One day = 24 hours = 86,400 seconds
dt = 86400

# Sea ice concentration threshold
sic_threshold = 0.15

# Starting day of the oil spill:
# NEMO files start on June 1st.
# November 20th = day 172.
START_DAY = 172

# Number of ensemble members
N_MEMBERS = 10

# Number of ensemble members that can be used to continue the simulation
# 1 = only the original member, until May 31st
# 10 = continue with following members until all beads melt or all members are used
MAX_MEMBERS_TO_CHAIN = 10


"""
IDW INTERPOLATION FUNCTION
"""

def idw(position_x, position_y, field):

    # Take lower integer coordinates around the particle
    x_int = int(np.floor(position_x))
    y_int = int(np.floor(position_y))

    # 4 surrounding grid points
    neighbors = [
        (x_int, y_int),
        (x_int + 1, y_int),
        (x_int, y_int + 1),
        (x_int + 1, y_int + 1)
    ]

    weights = []
    values = []

    # Loop through neighboring points
    for nx, ny in neighbors:

        # Check if point is inside the grid
        if 0 <= nx < field.shape[1] and 0 <= ny < field.shape[0]:

            value = field[ny, nx]

            # Skip masked or NaN values
            if np.ma.is_masked(value):
                continue

            if np.isnan(value):
                continue

            # Distance² between particle and grid point
            # +1e-12 avoids division by zero
            dist2 = (
                (position_x - nx)**2
                + (position_y - ny)**2
                + 1e-12
            )

            # IDW weight = 1 / distance²
            weights.append(1 / dist2)

            # Value of field at neighboring point
            values.append(value)

    # If no valid neighbor exists, return NaN
    if len(values) == 0:
        return np.nan

    # Weighted average
    return (
        np.sum(np.array(weights) * np.array(values))
        / np.sum(weights)
    )


"""
GLOBAL PROBABILITY MAP
"""

# This map counts how many ensemble members touch each grid cell.
# At the end, we divide by 10 to get a probability between 0 and 1.
prob_map = np.zeros(LON.shape, dtype=float)


"""
LOOP OVER THE 10 MEMBERS
"""

for k in range(N_MEMBERS):

    print("Running probability simulation for membre", k + 1)

    """
    OIL SPILL OF 100 OIL BEADS
    """

    # Make random result reproducible
    np.random.seed(42)

    # Oil beads start from the same starting point
    # but with a small position deviance inside the starting cell
    x = x0 + np.random.uniform(0, 1, N)
    y = y0 + np.random.uniform(0, 1, N)

    """
    IMPACT MAP FOR THIS MEMBER ONLY
    """

    # False = cell not touched by this member
    # True = cell touched by at least one oil bead in this member
    impact_this_member = np.zeros(LON.shape, dtype=bool)


    """
    TRAJECTORY WITH POSSIBLE CONTINUATION INTO NEXT MEMBERS
    """

    for chain in range(MAX_MEMBERS_TO_CHAIN):

        # Select which member is used for this part of the trajectory
        # Example:
        # k = 0 and chain = 0 -> member 1
        # k = 0 and chain = 1 -> member 2
        # k = 9 and chain = 1 -> member 1 again
        member_index = (k + chain) % N_MEMBERS

        print("Using data from membre", member_index + 1, "for trajectory of membre", k + 1)

        """
        OPEN NETCDF FILES
        """

        sic_ds = Dataset(sic_files[member_index])
        u_ds = Dataset(u_files[member_index])
        v_ds = Dataset(v_files[member_index])

        # Do not use [:] here, it is lighter for memory
        sic = sic_ds.variables["siconc"]
        u = u_ds.variables["sivelu"]
        v = v_ds.variables["sivelv"]

        number_of_days = sic.shape[0]

        # For the original member, start on November 20th.
        # For the following members, start on June 1st.
        if chain == 0:
            first_day = START_DAY
        else:
            first_day = 0


        """
        TIME LOOP
        """

        for t in range(first_day, number_of_days - 1):

            # Variables for current day
            u_t = u[t]
            v_t = v[t]
            sic_t = sic[t]


            """
            LOOP OVER OIL BEADS
            """

            for i in range(N):

                # Ignore removed particles
                if np.isnan(x[i]) or np.isnan(y[i]):
                    continue


                """
                INTERPOLATE VELOCITY
                """

                ui = idw(x[i], y[i], u_t)
                vi = idw(x[i], y[i], v_t)

                # Remove invalid interpolation results
                if np.isnan(ui) or np.isnan(vi):
                    x[i] = np.nan
                    y[i] = np.nan
                    continue


                """
                MOVE PARTICLE
                """

                x[i] += ui * dt / grid_size_m
                y[i] += vi * dt / grid_size_m

                # Safety check after movement
                if np.isnan(x[i]) or np.isnan(y[i]):
                    continue


                """
                CHECK IF PARTICLE IS STILL INSIDE GRID
                """

                ny, nx = sic_t.shape

                xi = int(round(x[i]))
                yi = int(round(y[i]))

                if (
                    xi < 0
                    or yi < 0
                    or xi >= nx
                    or yi >= ny
                ):

                    x[i] = np.nan
                    y[i] = np.nan
                    continue


                """
                MARK IMPACTED CELL
                """

                # This is the key point for a real probabilistic map:
                # the cell is marked True if at least one bead reaches it
                # during this member simulation.
                impact_this_member[yi, xi] = True


                """
                CHECK SEA ICE CONCENTRATION
                """

                sic_i = idw(x[i], y[i], sic_t)

                # Remove particle if sea ice concentration is below threshold
                if sic_i < sic_threshold:
                    x[i] = np.nan
                    y[i] = np.nan
                    continue


            """
            STOP IF ALL PARTICLES REMOVED
            """

            if np.all(np.isnan(x)):
                break


        """
        CLOSE FILES
        """

        sic_ds.close()
        u_ds.close()
        v_ds.close()

        # If all particles melted, stop using additional members
        if np.all(np.isnan(x)):
            break


    """
    ADD THIS MEMBER TO GLOBAL PROBABILITY MAP
    """

    # Each member counts only once per grid cell.
    # If a cell was touched in this member, add 1.
    prob_map += impact_this_member.astype(float)


"""
NORMALISATION
"""

# True probabilistic forecast:
# probability = number of impacted members / total number of members
prob_map = prob_map / N_MEMBERS


"""
PLOT PROBABILITY MAP
"""

fig = plt.figure(figsize=(10, 10))

ax = plt.axes(projection=ccrs.NorthPolarStereo())


"""
CARTE DE PROBABILITE
"""

pcm = ax.pcolormesh(
    LON,
    LAT,
    prob_map,
    transform=ccrs.PlateCarree(),
    cmap="YlOrRd",
    shading="auto",
    vmin=0,
    vmax=1,
    alpha=0.85
)

plt.colorbar(
    pcm,
    ax=ax,
    shrink=0.7,
    label="Probability of oil spill impact"
)


"""
CONTINENTS
"""

ax.coastlines()

ax.add_feature(
    cfeature.LAND,
    facecolor="lightgray"
)

ax.add_feature(
    cfeature.OCEAN,
    facecolor="white"
)


"""
POINT DE DEPART
"""

ax.scatter(
    LON[y0, x0],
    LAT[y0, x0],
    color="black",
    s=70,
    transform=ccrs.PlateCarree(),
    label="Starting point"
)

ax.legend()


"""
LIMITE ARCTIQUE
"""

ax.set_extent(
    [-180, 180, 60, 90],
    crs=ccrs.PlateCarree()
)

plt.title("Probabilistic oil spill impact forecast")

plt.savefig(
    "/work/figures3/probabilistic_oil_spill_forecast.png",
    dpi=300
)

plt.show()
plt.close()


"""
SAVE PROBABILITY MAP AS NUMPY FILE
"""

np.save(
    "/work/figures3/probabilistic_oil_spill_forecast.npy",
    prob_map
)

print("Probability map saved.")
```
